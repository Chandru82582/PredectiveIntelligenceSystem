```python
column_mapping = {

'datetime': 'event_datetime',

'CellID': 'cell_id',

'countrycode': 'country_code',

'smsin': 'sms_in_count',

'smsout': 'sms_out_count',

'callin': 'call_in_count',

'callout': 'call_out_count',
'internet': 'internet_usage'
}

canonical_activity = df.copy()
canonical_activity.columns = canonical_activity.columns.map(column_mapping)
```

```python
canonical_activity.head()
```

```python
canonical_activity["event_datetime"] = pd.to_datetime(canonical_activity["event_datetime"])
```

```python
canonical_activity.dtypes
```

```python
canonical_activity["event_datetime"].describe()
```

```python
canonical_activity["day_name"] = canonical_activity["event_datetime"].dt.day_name()
canonical_activity["date"] = canonical_activity["event_datetime"].dt.date
canonical_activity["hour"] = canonical_activity["event_datetime"].dt.hour 
```
