```python
column_mapping = {

'datetime': 'timestamp',

'CellID': 'grid_id',

'countrycode': 'country_code',

'smsin': 'sms_in_count',

'smsout': 'sms_out_count',

'callin': 'call_in_count',

'callout': 'call_out_count',
'internet': 'internet_usage'
}

canonical_activity = df.copy()
canonical_activity.columns = canonical_activity.columns.map(column_mapping)
```

```python
country_activity = (
canonical_activity.groupby(['timestamp', 'grid_id'])['country_code']
      .count()
      .reset_index(name='country_rows')
      .sort_values('country_rows', ascending=False)
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

# Count unique country codes for each grid
unique_country_grid = (
    canonical_activity
    .groupby('grid_id')['country_code']
    .nunique()
    .reindex(range(1, 10001), fill_value=0)
)

# Convert grid IDs to 100 × 100 positions
country_grid = unique_country_grid.values.reshape(100, 100)

# Plot separate heatmap
plt.figure(figsize=(12, 10))

sns.heatmap(
    country_grid,
    cmap='YlOrRd',
    xticklabels=False,
    yticklabels=False,
    cbar_kws={'label': 'Unique Country Count'}
)

plt.title('Unique Country Count Across 100 × 100 Grid')
plt.xlabel('Grid Column')
plt.ylabel('Grid Row')

plt.tight_layout()
plt.show()
```

```python
act_df = (
    canonical_activity
    .groupby([ 'grid_id', 'country_code'])
    .size()
    .reset_index(name='row_count')
)
```

```python
act_df['row_count'].nunique
```

```python
activity_metrics = [
    'sms_in_count',
    'sms_out_count',
    'call_in_count',
    'call_out_count',
    'internet_usage'
]

activity_by_grid_country = (
    canonical_activity
    .groupby(['grid_id', 'country_code'], as_index=False)[activity_metrics]
    .sum()
)
```

```python
activity_by_grid_country
```

```python
canonical_activity['total_sms'] = (
    canonical_activity['sms_in_count'] +
    canonical_activity['sms_out_count']
)

canonical_activity['total_calls'] = (
    canonical_activity['call_in_count'] +
    canonical_activity['call_out_count']
)

canonical_activity['total_activity'] = (
    canonical_activity['sms_in_count'] +
    canonical_activity['sms_out_count'] +
    canonical_activity['call_in_count'] +
    canonical_activity['call_out_count'] +
    canonical_activity['internet_usage']
)
```

```python
canonical_activity.head()
```

```python
# Sum activity for each grid
grid_activity = (
    canonical_activity
    .groupby('grid_id')[[
        'total_sms',
        'total_calls',
        'internet_usage',
        'total_activity'
    ]]
    .sum()
)

# Make sure all 10,000 grids exist
grid_activity = grid_activity.reindex(range(1, 10001), fill_value=0)

# Convert grid_id to 100 x 100 coordinates
grid_activity['row'] = (grid_activity.index - 1) // 100
grid_activity['col'] = (grid_activity.index - 1) % 100

# Create 100 x 100 matrices
sms_grid = grid_activity.pivot(
    index='row',
    columns='col',
    values='total_sms'
)

calls_grid = grid_activity.pivot(
    index='row',
    columns='col',
    values='total_calls'
)

internet_grid = grid_activity.pivot(
    index='row',
    columns='col',
    values='internet_usage'
)

total_grid = grid_activity.pivot(
    index='row',
    columns='col',
    values='total_activity'
)

# --------------------------------
# Plot 4 heatmaps
# --------------------------------

fig, axes = plt.subplots(2, 2, figsize=(18, 16))

heatmaps = [
    (sms_grid, 'Total SMS'),
    (calls_grid, 'Total Calls'),
    (internet_grid, 'Internet Usage'),
    (total_grid, 'Total Activity')
]

for ax, (data, title) in zip(axes.flat, heatmaps):

    sns.heatmap(
        data,
        ax=ax,
        cmap='YlOrRd',
        xticklabels=False,
        yticklabels=False,
        cbar_kws={'label': 'Activity'}
    )

    ax.set_title(title)
    ax.set_xlabel('Grid Column')
    ax.set_ylabel('Grid Row')

plt.suptitle(
    'Telecom Activity Across 100 × 100 Grid',
    fontsize=18
)

plt.tight_layout()
plt.show()
```
